# RazorGuard AI - Exploratory Data Analysis (EDA)
## Track 02 — Explainable AI Risk Manager for Payment Fraud

This notebook performs comprehensive exploratory analysis on the **100,000 payment transaction dataset**.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('dark_background')
df = pd.read_csv('../data/processed/processed_fraud_transactions.csv')
print(f'Dataset Shape: {df.shape}')
df.head()
```

### 1. Data Integrity & Class Distribution

```python
print('Missing Values:', df.isnull().sum().sum())
print('Duplicate Tx IDs:', df['transaction_id'].duplicated().sum())
print('\nClass Distribution:')
print(df['is_fraud'].value_counts(normalize=True).map('{:.2%}'.format))
```

### 2. Transaction Amount Dynamics (Legit vs Fraud)

```python
df.groupby('is_fraud')['amount'].describe()
```

### 3. Top Feature Correlations with Fraud Target

```python
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()['is_fraud'].sort_values(ascending=False)
print(corr)
```
